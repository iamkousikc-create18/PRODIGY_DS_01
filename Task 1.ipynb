```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

# Task 1: Data Visualization

## Objective

create a bar chart or histogram to visualize the distribution of a categorical or continuous variable, such as the distribution of ages or genders within a population dataset.

```python
df=pd.read_csv("API_SP.POP.TOTL_DS2_en_csv_v2_38144.csv",skiprows=4)
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
population_2022=df["2022"].dropna()
```

```python
population_2022.isnull().sum()
```

```python
plt.figure(figsize=(8,5))
plt.hist(np.log10(population_2022),bins=30)
plt.title("Distribution of Population in 2022")
plt.xlabel("Log10(Population)")
plt.ylabel("Number of Countries")
plt.show()
```

```python
top10=df[["Country Name","2022"]].dropna()
top10=top10.sort_values(by="2022",ascending=False).head(10)
```

```python
top10
```

```python
plt.figure(figsize=(16,6))
plt.bar(top10["Country Name"],top10["2022"])
plt.xticks(rotation=45)
plt.title("Top 10 Most Populated Countries")
plt.xlabel("Country")
plt.ylabel("Population")
plt.show()
```

# Insights

- The population data is highly right-skewed on a normal scale.
- A few countries have extremely large populations compared to most countries.
- Log transformation was applied to better visualize the distribution.
- After applying log scale, the population distribution appears more balanced and easier to interpret.
- This shows significant inequality in population sizes across countries.
